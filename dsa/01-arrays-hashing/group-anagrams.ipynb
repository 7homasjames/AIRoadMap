## Problem

Given an array of strings strs, group the anagrams together. You can return the answer in any order.

## Constraints

- 1 <= len(strs) <= 10^4
- 0 <= len(strs[i]) <= 100
- strs[i] consists of lowercase English letters.

## Examples

strs = ["eat","tea","tan","ate","nat","bat"] -> [["bat"],["nat","tan"],["ate","eat","tea"]]

strs = [""] -> [[""]]

strs = ["a"] -> [["a"]]

**Restated:** put words made of exactly the same letters into the same group.

## Step 2 (0:05-0:15): Approach — no coding yet.

Describe your approach in plain English or pseudocode. What's the brute force?

**My approach:** for each word not already grouped, compare it with every later word by sorting both. Matches go into its group, and their indices are marked as used.


## Step 3 (0:15-0:45): Code it.

```python
class Solution:
    def groupAnagrams(self, strs: list[str]) -> list[list[str]]:
        groups = []
        used = []

        for i in range(len(strs)):
            if i in used:
                continue

            group = [strs[i]]

            for j in range(i + 1, len(strs)):
                if sorted(strs[i]) == sorted(strs[j]):
                    group.append(strs[j])
                    used.append(j)

            groups.append(group)

        return groups
```

## Step 4 (0:45-0:55): Test with edge cases.

Add test cases below: the given examples, plus empty input, single element,
duplicates, negatives (whichever apply to this problem). Fix bugs here.

```python
import random, string, time

solve = Solution().groupAnagrams
norm = lambda groups: sorted(sorted(g) for g in groups)   # LeetCode accepts any order

# given examples
assert norm(solve(["eat", "tea", "tan", "ate", "nat", "bat"])) == norm([["bat"], ["nat", "tan"], ["ate", "eat", "tea"]])
assert solve([""]) == [[""]]
assert solve(["a"]) == [["a"]]

# edge cases
assert norm(solve(["", ""])) == [["", ""]]          # two empty strings are anagrams
assert norm(solve(["abc", "abc"])) == [["abc", "abc"]]  # exact duplicates stay in one group
assert norm(solve(["ab", "ba", "abc"])) == [["ab", "ba"], ["abc"]]  # different lengths never match

# speed: n = 1,000 random strings of length 1-100
random.seed(1)
big = ["".join(random.choices(string.ascii_lowercase, k=random.randint(1, 100))) for _ in range(1_000)]
t = time.time()
solve(big)
print(f"n=1,000 took {time.time() - t:.1f}s")
print("All tests passed")
```

- **Final complexity:** O(n² · k log k) time, O(n) space, where k is the word length. Every pair of words is compared, and each comparison sorts both words. `used` is a list, so `i in used` is another O(n) scan per word.
- **Would it pass LeetCode?** Probably not. 1,000 words took 2 s, and the limit is 10,000 words: 100× the pairs, so about 200 s. It will hit Time Limit Exceeded on the large tests.

**Faster approach, to try at the revisit (by 2026-10-31):** sort each word once and use the sorted word as a dictionary key, so every anagram lands in the same bucket in a single pass. That's O(n · k log k). The same idea as Two Sum: a hash map replaces the inner loop.

**Quick fixes if keeping this version:** make `used` a set, so `i in used` is O(1), and sort `strs[i]` once outside the inner loop instead of on every comparison. It's still O(n²) overall, though.
