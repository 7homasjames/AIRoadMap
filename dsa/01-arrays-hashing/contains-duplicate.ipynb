## Problem

Given an integer array nums, return True if any value appears at least twice in the array, and return False if every element is distinct.

## Constraints

- 1 <= len(nums) <= 10^5
- -10^9 <= nums[i] <= 10^9

## Examples

nums = [1, 2, 3, 1] -> True

nums = [1, 2, 3, 4] -> False

nums = [1, 1, 1, 3, 3, 4, 3, 2, 4, 2] -> True

**Restated:** given a list of integers, answer one yes/no question: does any value appear more than once?

**Constraints that matter:**
- `n` can be up to 100,000, so an O(n²) "compare every pair" approach (~5 billion comparisons) is too slow. Aim for O(n).
- Values range from -10⁹ to 10⁹, too wide for a fixed-size counting array. Use a hash set instead.
- The list always has at least 1 element (no empty input).

**Edge cases:**
- Single element, e.g. `[5]`: nothing to repeat, so `False`.
- Duplicate right at the start, e.g. `[1, 1, 2, 3]`: should return early.
- Duplicate only at the very end, e.g. `[1, 2, 3, 1]`: worst case, every element is checked.
- Negative numbers, e.g. `[-1, -1]`: work the same as positives in a set.
- All elements the same, e.g. `[7, 7, 7]`: `True`.

**Brute force:** compare every pair `(i, j)`. Time O(n²), space O(1). Too slow for n = 10⁵.

**Sort first:** sort, then check neighbours (`nums[i] == nums[i+1]`). Time O(n log n), space O(1) to O(n) depending on the sort.

**Approach 1: set length.** `set(nums)` keeps one copy of each value. If `len(set(nums)) < len(nums)`, something was removed, so there was a duplicate. Time O(n), space O(n). Always processes the whole list.

**Approach 2: seen-set with early exit (chosen).**
```
seen = empty set
for each number in nums:
    if number in seen  -> return True   (stop immediately)
    add number to seen
return False
```
Time O(n) worst case, space O(n). Stops as soon as the first duplicate appears.

**Why a set and not a list for `seen`?** `x in set` is O(1) on average (hashing). `x in list` is O(n), which would make the whole thing O(n²).

**Trick to remember:**
1. "Have I seen this before?" → use a **hash set**. Membership check is O(1), versus O(n) for a list.
