## Problem

Given an integer array nums and an integer k, return the k most frequent elements. You may return the answer in any order.

## Constraints

- 1 <= len(nums) <= 10^5
- -10^4 <= nums[i] <= 10^4
- k is in the range [1, number of unique elements]
- The answer is guaranteed to be unique.

## Examples

nums = [1,1,1,2,2,3], k = 2 -> [1,2]

nums = [1], k = 1 -> [1]

**My approach:** count each number in a dictionary, sort the distinct numbers by count from highest to lowest, and take the first k.


## Step 3 (0:15-0:45): Code it.

```python
class Solution:
    def topKFrequent(self, nums: list[int], k: int) -> list[int]:
        freq = {}

        for num in nums:
            freq[num] = freq.get(num, 0) + 1

        sorted_nums = sorted(freq, key=freq.get, reverse=True)

        return sorted_nums[:k]
```

```python
import random, time

solve = Solution().topKFrequent

# given examples (LeetCode accepts any order, so compare as sets)
assert set(solve([1, 1, 1, 2, 2, 3], 2)) == {1, 2}
assert solve([1], 1) == [1]

# edge cases
assert set(solve([4, 4, 5, 5, 5, 6], 1)) == {5}          # single most frequent
assert set(solve([-1, -1, 2], 1)) == {-1}                # negatives
assert set(solve([7, 8, 9], 3)) == {7, 8, 9}             # k = number of distinct values
assert set(solve([3, 0, 1, 0], 1)) == {0}                # zero as a value
assert len(solve([1, 2, 2, 3, 3, 3], 2)) == 2            # returns exactly k items

# speed: n = 100,000 (the maximum), values from -10^4 to 10^4
random.seed(1)
big = [random.randint(-10_000, 10_000) for _ in range(100_000)]
t = time.time()
solve(big, 100)
print(f"n=100,000 took {time.time() - t:.2f}s")
print("All tests passed")
```

- **Final complexity:** O(n + m log m) time, O(m) space, where m is the number of distinct values (at most n). Counting is O(n), and sorting the m distinct values is O(m log m).
- **Would it pass LeetCode?** Yes. 100,000 numbers took well under a second.

**Trick to remember:** a frequency map (`dict.get(x, 0) + 1`, or `collections.Counter`) turns "most common" questions into a sort over the counts. `sorted(freq, key=freq.get, reverse=True)` sorts the keys by their values.

**Going further (the follow-up asks for better than O(n log n)):**
- **Bucket sort, O(n):** make `buckets[c]` = the list of numbers that appear exactly c times. Counts are at most n, so there are n + 1 buckets. Walk the buckets from n down to 1 until you have k numbers. No sort needed.
- **Heap, O(n log k):** `heapq.nlargest(k, freq, key=freq.get)`, which keeps only k items in the heap.
