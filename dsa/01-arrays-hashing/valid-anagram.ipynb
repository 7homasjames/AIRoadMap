## Problem

Given two strings s and t, return true if t is an anagram of s, and false otherwise.

## Constraints

1 <= s.length, t.length <= 5 * 10^4; both strings consist of lowercase English letters.

## Examples

s = 'anagram', t = 'nagaram' -> true

s = 'rat', t = 'car' -> false

**Constraints (restated):** both strings are 1 to 50,000 lowercase letters. No uppercase, no unicode, no digits/symbols to worry about.

**Quick O(1) rejection:** if `len(s) != len(t)`, they can't be anagrams — no need to look at characters at all.

**Edge cases:**
- Equal-length strings that are *not* anagrams (e.g. `"aacc"` vs `"ccac"` — wait, that one *is* an anagram; a real non-example: `"aabb"` vs `"abbc"`)
- Identical strings (`s == t`) — trivially an anagram of itself
- Single-character strings (`"a"` vs `"a"` → true, `"a"` vs `"b"` → false)
- Strings with repeated letters, where counts must match exactly (`"aab"` vs `"aba"` → true, `"aab"` vs `"abb"` → false)


## Step 2 (0:05-0:15): Approach — no coding yet.

Describe your approach in plain English or pseudocode. What's the brute force?
Is there a better one? What's the time/space complexity of each?

**Brute force:** sort both strings and compare for equality. If they're anagrams, sorting produces the same sequence of characters for both. Time: O(n log n) for the sort, space O(n) for the sorted copies (or O(log n) auxiliary if the sort is in-place-ish).

**Better approach — frequency counting:** an anagram means every letter appears the same number of times in both strings. So:
1. If `len(s) != len(t)`, return False immediately (O(1) rejection).
2. Count the occurrences of each character in `s` into a hash map (or a fixed-size array of 26 counters, since we know it's lowercase a-z).
3. Walk through `t`, decrementing the count for each character.
4. If any count goes negative, or a character in `t` isn't in the map at all, return False.
5. If we get through all of `t` without a negative count, they're anagrams.

Time: O(n) — one pass to build counts, one pass to check. Space: O(1) technically (fixed 26-slot array, not O(n)) since the alphabet size is constant — this is the key improvement over sorting.


## Step 3 (0:15-0:45): Code it.

```python
def solution(s: str, t: str) -> bool:
    """
    LeetCode: Valid Anagram
    Pattern: Arrays Hashing
    Date: 2026-09-18
    Solved unaided: No (assistant-provided at user's request)
    Time complexity: O(n)
    Space complexity: O(1) (fixed 26-slot counter, alphabet size is constant)
    """
    if len(s) != len(t):
        return False

    counts = [0] * 26
    a = ord("a")

    for ch in s:
        counts[ord(ch) - a] += 1
    for ch in t:
        idx = ord(ch) - a
        counts[idx] -= 1
        if counts[idx] < 0:
            return False

    return True
```
