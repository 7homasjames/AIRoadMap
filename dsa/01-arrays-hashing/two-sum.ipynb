## Problem

Given an array of integers nums and an integer target, return the indices of the two numbers that add up to target. Each input has exactly one solution, and you may not use the same element twice. Return the answer in any order.

## Constraints

- 2 <= len(nums) <= 10^4
- -10^9 <= nums[i] <= 10^9
- -10^9 <= target <= 10^9
- Exactly one valid answer exists.

## Examples

nums = [2, 7, 11, 15], target = 9 -> [0, 1]

nums = [3, 2, 4], target = 6 -> [1, 2]

nums = [3, 3], target = 6 -> [0, 1]

**Restated:** find the two positions whose values add up to `target`, and return their indices, not the values.

**Constraints that matter:**
- `n` is up to 10,000, so O(n²) is about 50 million pair checks. That passes, but it's slow. Aim for O(n).
- Values can be negative and very large, so use a hash map rather than a counting array.
- An answer always exists, so there's no "not found" case to handle.

**Edge cases:**
- The same value twice, e.g. `[3, 3]`, target 6: two *different* indices with equal values.
- Using one element twice, e.g. `[3, 2, 4]`, target 6: `3 + 3` would match but isn't allowed.
- Negatives and zero, e.g. `[-3, 4, 3, 90]`, target 0.
- Unsorted input, so two pointers won't work unless you sort and keep track of the original indices.


## Step 2 (0:05-0:15): Approach — no coding yet.

**Brute force:** check every pair `(i, j)` with `i < j`. Time O(n²), space O(1).

**Sort + two pointers:** O(n log n), but sorting loses the original indices, so you'd have to carry `(value, index)` pairs. Messier than it's worth here.

**One-pass hash map (chosen):**
```
index_of = empty map        # value -> index
for i, x in nums:
    need = target - x
    if need in index_of -> return [index_of[need], i]
    index_of[x] = i
```
Time O(n), space O(n).

**Why check before storing?** If you stored `x` first, `[3, 2, 4]` with target 6 would find `3` itself at index 0 and return `[0, 0]`. Looking up before inserting means `x` can only pair with *earlier* elements, which also handles `[3, 3]` correctly.


## Step 3 (0:15-0:45): Code it.

````python
def solution(nums, target):
    """
    LeetCode: Two Sum
    Pattern: Arrays & Hashing
    Date: 2026-10-27
    Solved unaided: N (notebook written by assistant at user's request)
    Time complexity: O(n) - one pass, dict lookup/insert is O(1) average
    Space complexity: O(n) - the dict can hold every element before a match
    """
    index_of = {}                     # value -> index where we saw it
    for i, x in enumerate(nums):
        need = target - x
        if need in index_of:          # check BEFORE storing x, so x can't pair with itself
            return [index_of[need], i]
        index_of[x] = i
    return []                         # unreachable: exactly one solution is guaranteed


def solution_brute(nums, target):
    """Brute force: try every pair. O(n^2) time, O(1) space."""
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []
````
